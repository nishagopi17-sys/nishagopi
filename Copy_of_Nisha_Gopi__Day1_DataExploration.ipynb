```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Maths": [45, 60, 55, 70, 80, 65, 75, 90, 85, 95],
    "Science": [50, 65, 60, 75, 85, 70, 80, 95, 90, 100],
    "English": [40, 55, 50, 65, 75, 60, 70, 85, 80, 90]
}

df = pd.DataFrame(data)

print(df.head())
print("Shape:", df.shape)
print(df.isnull().sum())

sns.boxplot(data=df)

plt.title("Student Marks Analysis")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
```
